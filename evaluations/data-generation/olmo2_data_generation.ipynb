## OLMo‑2-1B-Instruct with Hugging Face

### This notebook loads allenai/OLMo‑2-1B-Instruct locally from huggingface and runs generation on Rubin data prompts.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, GenerationMixin
import torch
```

```python
model_name = "allenai/OLMo-2-0425-1B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
olmo = AutoModelForCausalLM.from_pretrained(model_name, trust_remote_code=True, device_map="auto")

device = torch.device("cpu")
olmo = olmo.to(device)
```

```python
def generate_text(prompt, max_new_tokens=1024, temperature=0.7, top_k=50, top_p=0.95):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    outputs = olmo.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=True,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p
    )
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
rubin_prompt = (
    "Using Rubin Observatory data, what are the key difficulties in classifying galaxies "
    "based on brightness, redshift, and image resolution?"
)

response = generate_text(rubin_prompt)
print("Rubin Response:\n", response)
```
